```python
import numpy as np
from scipy.stats import norm

epsilon = 0.15

regions = [
    (0.5, 0.2),
    (-0.4, 0.3),
    (0.1, 0.5)
]

def compute_probs(mu, sigma):
    p1 = 1 - norm.cdf(epsilon, mu, sigma)
    p0 = norm.cdf(-epsilon, mu, sigma)
    pNA = 1 - p1 - p0
    return p0, p1, pNA

for i, (mu, sigma) in enumerate(regions):
    p0, p1, pNA = compute_probs(mu, sigma)
    entropy = -sum(p*np.log(p) for p in [p0, p1, pNA] if p > 0)

    print(f"Region {i+1}")
    print(f"Probabilities: 0={p0:.2f}, 1={p1:.2f}, NA={pNA:.2f}")
    print(f"Entropy: {entropy:.3f}\n")
```
