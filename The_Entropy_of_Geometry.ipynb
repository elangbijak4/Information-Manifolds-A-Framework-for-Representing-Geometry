```python
import numpy as np

epsilon = 0.15

R = np.array([
    [0.8, 0.2, -0.5],
    [0.1, 0.0, -0.3],
    [0.6, -0.2, -0.7]
])

def encode(val):
    if val > epsilon:
        return 1
    elif val < -epsilon:
        return 0
    else:
        return "NA"

# Explicitly specify otypes=[object] to allow mixed types (int and string)
vectorized = np.vectorize(encode, otypes=[object])
X = vectorized(R)

print("Encoded:\n", X)

# Filter out 'NA' values before flattening and calculating entropy
# Entropy should only be calculated for the numerical outcomes (0 and 1)
filtered_flat = X[X != 'NA'].flatten()

# Check if there are any valid values to calculate entropy
if len(filtered_flat) == 0:
    print("Cannot calculate entropy: No valid numerical values (0 or 1) after filtering 'NA'.")
else:
    unique, counts = np.unique(filtered_flat, return_counts=True)

    # Calculate probabilities based on the filtered values
    probs = counts / len(filtered_flat)
    entropy = -np.sum(probs * np.log(probs))

    print("Entropy:", entropy)
```
